```
import math

SKILL_ALIASES = {
    "python": "python",
    "pyhton": "python",
    "java": "java",
    "javascript": "javascript",
    "javascrpit": "javascript",
    "js": "javascript",
    "typescript": "typescript",
    "typescrpit": "typescript",
    "c++": "cpp",
    "cpp": "cpp",
    "r": "r",
    "kotlin": "kotlin",
    "machinelearning": "machine_learning",
    "machine learning": "machine_learning",
    "ml": "machine_learning",
    "sklearn": "machine_learning",
    "deeplearning": "deep_learning",
    "deep learning": "deep_learning",
    "deep-learning": "deep_learning",
    "tensorflow": "tensorflow",
    "pytorch": "pytorch",
    "keras": "keras",
    "nlp": "nlp",
    "bert": "bert",
    "xgboost": "xgboost",
    "feature engineering": "feature_engineering",
    "statistics": "statistics",
    "stats": "statistics",
    "regression": "regression",
    "clustering": "clustering",
    "data-viz": "data_visualization",
    "data visualization": "data_visualization",
    "data viz": "data_visualization",
    "matplotlib": "data_visualization",
    "tableau": "data_visualization",
    "power-bi": "data_visualization",
    "power bi": "data_visualization",
    "powerbi": "data_visualization",
    "pandas": "pandas",
    "numpy": "numpy",
    "react": "react",
    "reacts": "react",
    "reactjs": "react",
    "vue": "vue",
    "vue.js": "vue",
    "vuejs": "vue",
    "redux": "redux",
    "tailwind": "tailwind",
    "html/css": "html_css",
    "html css": "html_css",
    "html": "html_css",
    "css": "html_css",
    "jest": "jest",
    "graphql": "graphql",
    "node.js": "nodejs",
    "nodejs": "nodejs",
    "node js": "nodejs",
    "flask": "flask",
    "spring boot": "spring_boot",
    "springboot": "spring_boot",
    "rest api": "rest_api",
    "rest": "rest_api",
    "restapi": "rest_api",
    "microservices": "microservices",
    "sql": "sql",
    "mysql": "mysql",
    "mysq": "mysql",
    "postgresql": "postgresql",
    "postgres": "postgresql",
    "mongodb": "mongodb",
    "redis": "redis",
    "docker": "docker",
    "kubernetes": "kubernetes",
    "kubernates": "kubernetes",
    "k8s": "kubernetes",
    "ci/cd": "ci_cd",
    "cicd": "ci_cd",
    "ci cd": "ci_cd",
    "aws": "aws",
    "android": "android",
    "firebase": "firebase",
    "algorithms": "algorithms",
    "algoritms": "algorithms",
    "data structure": "data_structures",
    "data structures": "data_structures",
    "competitive programming": "competitive_programming",
    "ui/ux": "ui_ux",
    "ui ux": "ui_ux",
    "figma": "figma",
}

candidates = {
    "Arjun Sharma": ["Pyhton", "MachineLearning", "SQL", "pandas", "numpy", "Deep-learning"],
    "Priya Nair": ["JavaScrpit", "React", "Node.JS", "MongoDb", "REST api", "HTML/CSS"],
    "Rahul Gupta": ["Java", "Spring Boot", "MySql", "Microservices", "Docker", "kubernates"],
    "Sneha Patel": ["Python", "TensorFlow", "Keras", "NLP", "BERT", "data-viz", "matplotlib"],
    "Vikram Singh": ["C++", "Algoritms", "Data Structure", "competitive programming", "python"],
    "Ananya Krishnan": ["javascript", "vue.js", "python", "flask", "PostgreSQL", "AWS", "CI/CD"],
    "Karan Mehta": ["Python", "Sklearn", "XGboost", "feature engineering", "SQL", "tableau"],
    "Deepika Rao": ["Java", "Android", "Kotlin", "Firebase", "REST", "UI/UX", "figma"],
    "Aditya Kumar": ["Reactjs", "TypeScrpit", "GraphQL", "redux", "tailwind", "nodejs", "jest"],
    "Meera Iyer": ["python", "R", "statistics", "ML", "regression", "clustering", "Power-BI"],
}

job_descriptions = {
    "JD-1": [
        "python", "machine learning", "deep learning", "tensorflow",
        "pytorch", "sql", "data visualization", "nlp", "bert",
        "feature engineering", "statistics"
    ],

    "JD-2": [
        "java", "spring boot", "mysql", "postgresql",
        "microservices", "docker", "kubernetes",
        "rest api", "ci/cd", "redis"
    ],

    "JD-3": [
        "javascript", "react", "vue", "typescript",
        "rest api", "html/css", "node.js",
        "graphql", "redux", "jest", "aws"
    ],
}


# Normalize skills function
def normalize_skills(skills_list):

    normalized_skills = []

    for skill in skills_list:

        skill = skill.lower().strip()

        if skill in SKILL_ALIASES:
            normalized_skills.append(SKILL_ALIASES[skill])

        else:
            skill = skill.replace("-", " ").replace("/", " ")

            if skill in SKILL_ALIASES:
                normalized_skills.append(SKILL_ALIASES[skill])

    return list(set(normalized_skills))


# Normalize candidate skills
normalized_candidates = {}

for candidate, skills in candidates.items():
    normalized_candidates[candidate] = normalize_skills(skills)


# Normalize job description skills
normalized_job_descriptions = {}

for job, skills in job_descriptions.items():
    normalized_job_descriptions[job] = normalize_skills(skills)


# Calculate TF-IDF
def calculate_tfidf(candidate_skills):

    tf_idf = {}

    total_candidates = len(normalized_candidates)

    for skill in candidate_skills:

        tf = 1 / len(candidate_skills)

        document_frequency = sum(
            1 for skills in normalized_candidates.values()
            if skill in skills
        )

        idf = math.log(total_candidates / document_frequency)

        tf_idf[skill] = tf * idf

    return tf_idf


# TF-IDF vectors for candidates
tfidf_candidates = {
    candidate: calculate_tfidf(skills)
    for candidate, skills in normalized_candidates.items()
}


# Create vocabulary
vocabulary = list(set(
    skill
    for skills in normalized_candidates.values()
    for skill in skills
))


# Binary vector for job descriptions
def calculate_binary_vector(job_skills, vocabulary):

    binary_vector = [
        1 if skill in job_skills else 0
        for skill in vocabulary
    ]

    return binary_vector


job_description_binary_vectors = {
    job: calculate_binary_vector(skills, vocabulary)
    for job, skills in normalized_job_descriptions.items()
}


# Cosine similarity
def calculate_cosine_similarity(tfidf_vector, binary_vector):

    tfidf_vector_values = [
        tfidf_vector.get(skill, 0)
        for skill in vocabulary
    ]

    dot_product = sum(
        a * b
        for a, b in zip(tfidf_vector_values, binary_vector)
    )

    tfidf_magnitude = math.sqrt(
        sum(a * a for a in tfidf_vector_values)
    )

    binary_magnitude = math.sqrt(
        sum(b * b for b in binary_vector)
    )

    if tfidf_magnitude == 0 or binary_magnitude == 0:
        return 0.0

    return dot_product / (tfidf_magnitude * binary_magnitude)


# Calculate similarities
cosine_similarities = {}

for job, binary_vector in job_description_binary_vectors.items():

    similarities = {}

    for candidate, tfidf_vector in tfidf_candidates.items():

        similarity = calculate_cosine_similarity(
            tfidf_vector,
            binary_vector
        )

        similarities[candidate] = similarity

    cosine_similarities[job] = similarities


# Print top candidates
for job, similarities in cosine_similarities.items():

    sorted_similarities = sorted(
        similarities.items(),
        key=lambda x: (-x[1], x[0])
    )

    print(f"\n{job} - Top 3 Candidates:")

    for candidate, similarity in sorted_similarities[:3]:
        print(f"{candidate}: {similarity:.2f}")

    print(f"\nAll Candidates for {job}:")

    for candidate, similarity in sorted_similarities:
        print(f"{candidate}: {similarity:.2f}")
```
